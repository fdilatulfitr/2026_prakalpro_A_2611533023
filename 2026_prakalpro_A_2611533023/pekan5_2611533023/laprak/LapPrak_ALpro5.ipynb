## 2.1 Dasar Teori
Struktur percabangan (conditional statement) adalah instruksi yang memungkinkan program memilih dan mengeksekusi blok kode tertentu berdasarkan hasil evaluasi ekspresi logika yang bernilai benar (True) atau salah (False). konsep-konsep yang digunakan waktu praktikum minggu keempat yaitu:

1. IF Tunggal
Struktur if tunggal merupakan bentuk percabangan paling dasar. Blok kode di dalam if hanya akan dieksekusi jika kondisi yang diuji bernilai True. Jika kondisi bernilai False, program akan mengabaikan blok tersebut dan melanjutkan ke baris kode berikutnya. Bentuk ini biasanya digunakan untuk memeriksa atau memvalidasi suatu kondisi khusus.

2. Struktur IF-ELSE
Struktur if-else digunakan ketika terdapat dua jalur alternatif. Jika kondisi pada if bernilai True, maka blok if yang dijalankan. Namun, jika kondisi bernilai False, alur eksekusi secara otomatis berpindah ke blok else.

3. Struktur IF-ELIF-ELSE dan Operator Logika
Struktur if-elif-else digunakan untuk menangani lebih dari dua kemungkinan kondisi yang bersifat saling lepas. Evaluasi dilakukan secara berurutan dari atas ke bawah. Ketika salah satu kondisi bernilai True, blok kode terkait akan dieksekusi dan sisa kondisi di bawahnya diabaikan.

4. Multi-IF Terpisah
Struktur multi-if terpisah terdiri dari beberapa pernyataan if independen yang ditulis secara sejajar. Berbeda dengan if-elif-else, setiap blok if pada struktur ini akan selalu dievaluasi satu per satu tanpa memedulikan hasil dari blok if sebelumnya. Struktur ini cocok digunakan untuk pengujian kondisi yang dapat terjadi secara bersamaan (misalnya akumulasi nilai atau kriteria diskon).

5. Struktur MATCH-CASE
Fitur match-case (dikenalkan pada Python 3.10) merupakan bentuk pattern matching yang mencocokkan nilai variabel terhadap beberapa pola (case). Struktur ini membuat kode lebih rapi dibandingkan rantai if-elif yang panjang. Penggunaan case _ berfungsi sebagai penangan nilai default apabila tidak ada pola yang cocok.

### 2.2.4 Program "nested_for1_2611533023.py"

### 2.2.7 Program "nested_for4_2611533023.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
tinggi_3023 = int(input("masukan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3023 % 2 != 0:
    print("tinggi harus bilangan genap!")
else: 
    a = tinggi_3023
    c = a
    lebar_3023 = (2 * tinggi_3023) - 2

    for i in range(1, tinggi_3023 + 1):
        b = c + 1

        for j in range(1, lebar_3023 + 1):

            #baris atas dan bawah
            if i == 1 or i == tinggi_3023:
                if j == 1 or j == lebar_3023:
                    print("#", end="")
                else:
                    print("=", end="")
            #baris isi
            else:
                if j == 1 or j == lebar_3023:
                    print("|", end="")
                else:
                    if j == c:
                        print("<", end="")
                    elif j == b:
                        print(">", end="")
                    elif j == (lebar_3023 - c):
                        print("<", end="")
                    elif j == (lebar_3023 - c + 1):
                        print(">", end="")
                    elif j > b and j < (lebar_3023 - c):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()
        # logika asli java
        a -= 2
        if a <= 0:
            c = (-a) + 2
        else:
            c = a
```

**Penjelasan :**
1. Pemeriksaan Input (if tinggi_3023 % 2 != 0:):
   Program meminta masukan angka tinggi pola. Jika angka yang dimasukkan ganjil, program mencetak pesan peringatan bahwa tinggi harus bilangan genap dan program tidak akan dilanjutkan.

2. Inisialisasi Variabel:
   Jika input genap, program menghitung lebar bingkai (lebar_3023) dan menyiapkan variabel penentu posisi karakter internal (a dan c).

3. Perulangan Baris (for i in range(1, tinggi_3023 + 1):):
   Mengatur alur pencetakan dari baris pertama sampai baris paling bawah.

4. Pembentukan Komponen Karakter (if-else di dalam loop):
   a. Bingkai Atas & Bawah (i == 1 atau i == tinggi_3023): Mencetak sudut # di ujung kiri dan kanan, serta karakter = di antaranya.
   b. Bingkai Tepi Kiri & Kanan: Mencetak karakter | di kolom pertama (j == 1) dan kolom terakhir.
   c. Isi Pola Internal: Mencetak pasangan simbol < dan > yang bergerak mendekat ke tengah lalu menjauh kembali (membentuk pola jam pasir/kristal), serta mengisi area tengah dengan karakter titik (.) dan sisa kolom dengan spasi.

5. Update Posisi (a -= 2 & penyesuaian c):
   Setiap kali satu baris selesai dicetak, nilai variabel posisi diubah agar simetri pola bagian atas dan bagian bawah terbentuk dengan presisi.

```python
# === PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===

print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_3023 = int(input("Masukkan ukuran skala jam pasir (N): "))

# --------------------------------------------------
# 1. Bingkai Pembatas Horizontal Atas
# --------------------------------------------------
print("#", end="")
for i_3023 in range(4 * n_3023 + 5):
    print("=", end="")
print("#")

# --------------------------------------------------
# 2. Fase 1: Jam Pasir Atas (Reduksi Angka Menurun: N turun s.d. 1)
# --------------------------------------------------
for baris_3023 in range(n_3023, 0, -1):
    print("| ", end="")
    
    # Spasi penyeimbang kiri
    for spasi_3023 in range(2 * (n_3023 - baris_3023)):
        print(" ", end="")
        
    # Deret angka mundur dari baris turun ke 1
    for angka_3023 in range(baris_3023, 0, -1):
        print(angka_3023, end="")
        print(" ", end="")
            
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju dari 1 naik ke baris
    for angka_3023 in range(1, baris_3023 + 1):
        print(" ", end="")
        print(angka_3023, end="")
        
    # Spasi penyeimbang kanan
    for spasi_3023 in range(2 * (n_3023 - baris_3023)):
        print(" ", end="")
        
    print(" |")

# --------------------------------------------------
# 3. Fase 2: Poros Titik Pusat Jam Pasir (Titik Nol / Singularity)
# --------------------------------------------------
print("|", end="")
for spasi_3023 in range(2 * n_3023 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_3023 in range(2 * n_3023 + 1):
    print(" ", end="")
print("|")

# --------------------------------------------------
# 4. Fase 3: Jam Pasir Bawah (Ekspansi Angka Menaik: 1 naik s.d. N)
# --------------------------------------------------
for baris_3023 in range(1, n_3023 + 1):
    print("| ", end="")
    
    # Spasi penyeimbang kiri
    for spasi_3023 in range(2 * (n_3023 - baris_3023)):
        print(" ", end="")
        
    # Deret angka mundur dari baris turun ke 1
    for angka_3023 in range(baris_3023, 0, -1):
        print(angka_3023, end="")
        print(" ", end="")
            
    # Poros kristal tengah
    print("<*>", end="")
    
    # Deret angka maju dari 1 naik ke baris
    for angka_3023 in range(1, baris_3023 + 1):
        print(" ", end="")
        print(angka_3023, end="")
        
    # Spasi penyeimbang kanan
    for spasi_3023 in range(2 * (n_3023 - baris_3023)):
        print(" ", end="")
        
    print(" |")

# --------------------------------------------------
# 5. Bingkai Pembatas Horizontal Bawah
# --------------------------------------------------
print("#", end="")
for i_3023 in range(4 * n_3023 + 5):
    print("=", end="")
print("#")
```

Berdasarkan seluruh rangkaian praktikum yang telah dilakukan, dapat disimpulkan bahwa struktur kontrol perulangan merupakan fondasi utama dalam logika pemrograman untuk mengeksekusi instruksi secara berulang dan efisien. Penggunaan perulangan for dengan bantuan fungsi range() mempermudah pengolahan deret angka, baik secara menaik, menurun, maupun dalam perhitungan akumulasi total nilai.
Selain perulangan tunggal, penerapan perulangan bersarang (nested for) terbukti sangat krusial dalam mengolah data dua dimensi. Melalui kombinasi perulangan luar untuk mengatur posisi baris dan perulangan dalam untuk mengatur posisi kolom, kita dapat memanipulasi ruang dan karakter dengan presisi. Penggabungan nested for dengan kondisi if-else memungkinkan pembentukan berbagai pola visual kompleks, mulai dari pola sederhana seperti persegi dan segitiga, hingga pola simetris tingkat lanjut seperti bingkai gambar dan jam pasir kristal. Pemahaman terhadap alur eksekusi perulangan ini sangat melatih logika berpikir sistematis dalam menyelesaikan permasalahan pemrograman dua dimensi.
