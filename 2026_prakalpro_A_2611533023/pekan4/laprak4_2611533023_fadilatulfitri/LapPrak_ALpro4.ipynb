## 2.1 Dasar Teori
Struktur percabangan (conditional statement) adalah instruksi yang memungkinkan program memilih dan mengeksekusi blok kode tertentu berdasarkan hasil evaluasi ekspresi logika yang bernilai benar (True) atau salah (False). konsep-konsep yang digunakan waktu praktikum minggu keempat yaitu:

1. IF Tunggal
Struktur if tunggal merupakan bentuk percabangan paling dasar. Blok kode di dalam if hanya akan dieksekusi jika kondisi yang diuji bernilai True. Jika kondisi bernilai False, program akan mengabaikan blok tersebut dan melanjutkan ke baris kode berikutnya. Bentuk ini biasanya digunakan untuk memeriksa atau memvalidasi suatu kondisi khusus.

2. Struktur IF-ELSE
Struktur if-else digunakan ketika terdapat dua jalur alternatif. Jika kondisi pada if bernilai True, maka blok if yang dijalankan. Namun, jika kondisi bernilai False, alur eksekusi secara otomatis berpindah ke blok else.

3. Struktur IF-ELIF-ELSE dan Operator Logika
Struktur if-elif-else digunakan untuk menangani lebih dari dua kemungkinan kondisi yang bersifat saling lepas. Evaluasi dilakukan secara berurutan dari atas ke bawah. Ketika salah satu kondisi bernilai True, blok kode terkait akan dieksekusi dan sisa kondisi di bawahnya diabaikan.

4. Multi-IF Terpisah
Struktur multi-if terpisah terdiri dari beberapa pernyataan if independen yang ditulis secara sejajar. Berbeda dengan if-elif-else, setiap blok if pada struktur ini akan selalu dievaluasi satu per satu tanpa memedulikan hasil dari blok if sebelumnya. Struktur ini cocok digunakan untuk pengujian kondisi yang dapat terjadi secara bersamaan (misalnya akumulasi nilai atau kriteria diskon).

5. Struktur MATCH-CASE
Fitur match-case (dikenalkan pada Python 3.10) merupakan bentuk pattern matching yang mencocokkan nilai variabel terhadap beberapa pola (case). Struktur ini membuat kode lebih rapi dibandingkan rantai if-elif yang panjang. Penggunaan case _ berfungsi sebagai penangan nilai default apabila tidak ada pola yang cocok.



## 2.2 Kode Program

### 2.2.1 Program "if_elif_else1_2611533023.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
umur_3023 = int(input("input umur anda: "))
sim_3023 = input("apakah anda sudah punya sim C: ")[0]

if umur_3023 >= 17 and sim_3023 == 'y':
    print("anda sudah dewasa dan boleh bawa motor")
elif umur_3023 >= 17 and sim_3023 != 'y':
    print("anda sudah dewasa tapi tidak boleh bawa motor")
else:
    print("anda belum cukup umur dan tidak boleh bawa motor")
print("program selesai")
```

**Penjelasan :**
1. Input Data: Program meminta umur dan status SIM. Kode [0] mengambil huruf pertama saja dari jawaban SIM (misal: 'ya' diambil 'y').
2. Kondisi 1 (if): Jika umur 17+ dan punya SIM ('y'), muncul pesan dewasa dan boleh bawa motor..
3. Kondisi 2 (elif): Jika umur 17+ tapi tidak punya SIM (!= 'y'), muncul pesan dewasa tapi dilarang bawa motor.
4. Kondisi 3 (else): Jika umur di bawah 17, otomatis muncul pesan belum cukup umur.
5. Baris Akhir: Menampilkan "program selesai" setelah pengecekan kondisi rampung.

### 2.2.4 Program "match_case1.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
bulan_3023 = int(input("masukan angka bulan (1 - 12): "))

match bulan_3023:
    case 1:
        print("januari")
    case 2:
        print("februari")
    case 3:
        print("maret")
    case 4:
        print("april")
    case 5:
        print("mei")
    case 6:
        print("juni")
    case 7:
        print("juli")
    case 8:
        print("agustus")
    case 9:
        print("september")
    case 10:
        print("oktober")
    case 11:
        print("november")
    case 12:
        print("desember")
    case _:
        print("angka tidak valid")
```

**Penjelasan :**
1. Input Data (int): Program meminta pengguna memasukkan angka dari 1 sampai 12, lalu mengonversinya menjadi bilangan bulat (int) untuk disimpan pada variabel bulan_3023.

2. Pencocokan Nilai (match-case): Program mengevaluasi nilai variabel bulan_3023 dengan mencocokkannya ke setiap pilihan kasus (case) yang tersedia.

3. Pilihan Kasus 1–12: Jika angka yang dimasukkan berada dalam rentang 1 hingga 12, program akan mencetak nama bulan yang sesuai (misalnya, angka 1 untuk "januari", 2 untuk "februari", dan seterusnya).

4. Kasus Default (case _): Jika angka yang dimasukkan di luar rentang 1–12 (misalnya angka 0 atau 13), program akan mengeksekusi simbol _ sebagai penanganan kondisi acuan dan mencetak pesan "angka tidak valid".




### 2.2.5 Program "multi_if1.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
umur_3023 = int(input("input umur anda: "))
sim_3023 = input("apakah anda sudah punya sim C (y/t): ")[0]

if umur_3023 >= 17 and sim_3023 == 'y':
    print("anda sudah dewasa dan boleh bawa motor")
if umur_3023 >= 17 and sim_3023 != 'y':
    print("anda sudah dewasa tetapi tidak boleh bawa motor")
if umur_3023 < 17 and sim_3023 != 'y':
    print("anda belum cukup umur bawa motor")
if umur_3023 < 17 and sim_3023 == 'y':
    print("anda belum cukup umur punya SIM")
```

**Penjelasan :**

2. Evaluasi Multi-IF Terpisah: Program menggunakan empat pernyataan if independen, di mana setiap baris akan diperiksa satu per satu secara berurutan:

if Pertama: Jika umur 17 tahun ke atas dan punya SIM ('y'), muncul pesan bahwa pengguna sudah dewasa dan boleh bawa motor.

if Kedua: Jika umur 17 tahun ke atas tetapi tidak punya SIM (!='y'), muncul pesan dewasa tetapi tidak boleh bawa motor.

if Ketiga: Jika umur di bawah 17 tahun dan tidak punya SIM (!='y'), muncul pesan belum cukup umur bawa motor.

if Keempat: Jika umur di bawah 17 tahun tetapi mengaku punya SIM ('y'), muncul pesan bahwa pengguna belum cukup umur untuk memiliki SIM.

3. Penyelesaian Transaksi: Karena seluruh kondisi bersifat saling lepas (mutually exclusive), hanya satu dari empat baris print di atas yang pesan tampilannya akan muncul di layar.



### 2.2.6 Program "multi_if2_2611533023.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# input dari user
total_belanja_3023 = float(input("masukan total belanja (Rp): "))

#input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_3023 = input("apakah anda member? (y/t): ").strip().lower()
is_member_3023 = input_member_3023 in ["y", "ya"]

#input status kode promo (mengecek apakah user mengetik 'y' ayau 'ya')
input_promo_3023 = input("apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_3023 = input_promo_3023 in ["y", "ya"]
total_diskon_persen_3023 = 0

#multi-if terpisah: setia kondisi diperiksa secara independen
#diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_3023 > 1000000:
    total_diskon_persen_3023 += 10 #diskon belanja besar

if is_member_3023:
    total_diskon_persen_3023 +=5 #diskon member

if kode_promo_valid_3023:
    total_diskon_persen_3023 +=15 #diskon voucher

#menghitung nominal diskon dan total bayar
nominal_diskon_3023 = total_belanja_3023 * (total_diskon_persen_3023 / 100)
total_bayar_3023 = total_belanja_3023 - nominal_diskon_3023

#output hasil
print("\n--- rincian pembayaran ---")
print(f"total diskon : {total_diskon_persen_3023}% (rp {nominal_diskon_3023:,.0f})")
print(f"total bayar :rp {total_bayar_3023:,.0f}")

print(f"total diskon yang anda dapatkan: {total_diskon_persen_3023}%")
# output: total diskon yang anda dapatkan: 30% jika belanja > 1 juta,member,dan kode promo valid
```

**Penjelasan :**
1. Input Data dan Format Teks: Program meminta input total belanja, status member, dan validasi kode promo. Penggunaan .strip().lower() dan ekspresi in ["y", "ya"] mengubah jawaban pengguna menjadi nilai logika (boolean) True jika pengguna menjawab "y" atau "ya", dan False untuk jawaban lainnya.

2. Akumulasi Diskon (Multi-IF Terpisah): Program menggunakan tiga pernyataan if independen untuk memeriksa kriteria diskon:

if Pertama: Jika total belanja di atas Rp1.000.000, diskon bertambah 10%.

if Kedua: Jika pengguna adalah member (True), diskon bertambah 5%.

if Ketiga: Jika kode promo valid (True), diskon bertambah 15%.

Karena setiap if berdiri sendiri, persentase diskon akan ditumpuk (diakumulasikan) jika pengguna memenuhi lebih dari satu syarat.

3. Kalkulasi Biaya: Program menghitung besaran nilai rupiah dari total persen diskon, lalu mengurangkannya dari total belanja untuk mendapatkan nominal pembayaran akhir.

4. Cetak Hasil: Program menampilkan rincian total diskon dan total bayar menggunakan format desimal ribuan (:,.0f).

Beberapa blok if independen digunakan pada bagian perhitungan diskon. Karena setiap kondisi berdiri sendiri, sistem memeriksa satu per satu kriteria (subtotal besar, status member, kode promo, dan jumlah tiket). Jika pengunjung memenuhi lebih dari satu kriteria, seluruh nilai diskonnya akan ditambahkan secara terakumulasi.
