## 2.1 Dasar Teori
Operator adalah simbol yang digunakan untuk melakukan suatu operasi terhadap nilai atau variabel dalam sebuah program. Nilai yang digunakan oleh operator disebut operand. Dalam Python terdapat beberapa jenis operator yang memiliki fungsi
berbeda-beda. konsep yang dipelajari dalam praktikum ini adalah sebagai berikut. konsep-konsep yang digunakan waktu praktikum minggu kedua yaitu:

1. Operator aritmatika digunakan untuk melakukan perhitungan. Operator ini hampir sama seperti operasi matematika yang biasa kita gunakan, seperti penjumlahan, pengurangan, perkalian, dan pembagian. Dalam Python terdapat operator + untuk penjumlahan, - untuk pengurangan, * untuk perkalian, / untuk pembagian, % untuk mencari sisa pembagian, ** untuk perpangkatan, dan // untuk pembagian bulat. Contohnya, jika kita memiliki a = 10 dan b = 3, maka a + b akan menghasilkan 13. Sedangkan a % b akan menghasilkan 1 karena sisa dari 10 dibagi 3 adalah 1. Operator aritmatika ini biasanya digunakan ketika program membutuhkan perhitungan.

2. Operator assignment digunakan untuk memberikan atau mengubah nilai pada sebuah variabel. Operator yang paling sering digunakan adalah "=". Misalnya kita menulis "nilai = 80", artinya variabel nilai memiliki nilai 80. Selain tanda "=", terdapat juga "+=", "-=", "*=", dan "/=" yang dapat digunakan untuk mengubah nilai yang sudah ada. Contohnya, jika "nilai = 80" kemudian ditulis "nilai += 5", maka nilai tersebut akan berubah menjadi 85. Operator ini banyak digunakan karena hampir setiap program membutuhkan variabel untuk menyimpan data.

3. Operator bitwise digunakan untuk melakukan operasi pada bilangan dalam bentuk biner. Jadi, operator ini bekerja pada angka dalam bentuk bit. Beberapa operator bitwise yang ada dalam Python adalah "&", "|", "^", "~", "<<", dan ">>". Masing-masing memiliki fungsi yang berbeda. Misalnya, "&" digunakan untuk operasi AND, sedangkan "|" digunakan untuk operasi OR. Ada juga "<<" dan ">>" yang digunakan untuk menggeser bit ke kiri atau ke kanan. Operator ini memang sedikit berbeda dari operator yang biasa digunakan untuk perhitungan, karena cara kerjanya berhubungan dengan bentuk biner suatu bilangan.

4. Operator pembandingan digunakan untuk membandingkan dua nilai. Operator ini terdiri dari "==", "!=", ">", "<", ">=", dan "<=". Hasil dari perbandingan biasanya berupa "True" atau "False". Misalnya, "80 >= 75" akan menghasilkan "True" karena 80 lebih besar dari 75. Sedangkan "60 >= 75" akan menghasilkan "False" karena 60 belum mencapai 75. Operator pembandingan sering digunakan ketika program harus menentukan suatu kondisi, misalnya untuk menentukan apakah seseorang lulus atau tidak.

5. Operator logika digunakan ketika kita ingin menggabungkan beberapa kondisi. Operator yang digunakan adalah "and", "or", dan "not". "And" digunakan jika semua kondisi harus benar, sedangkan "or" digunakan jika salah satu kondisi sudah benar. Sementara itu, "not" digunakan untuk membalikkan suatu kondisi. Contohnya, jika kita ingin membuat kondisi bahwa mahasiswa dinyatakan lulus jika nilainya minimal 75 dan kehadirannya memenuhi syarat, kita dapat menggunakan operator "and". Operator logika biasanya digunakan bersama dengan operator pembandingan.

6. operator lainnya, Salah satunya adalah operator "in" dan "not in" Selain itu, terdapat operator "is" dan "is not". Operator ini digunakan untuk melihat apakah dua objek merupakan objek yang sama atau tidak. Operator "is" berbeda dengan "==". Operator "==" digunakan untuk membandingkan nilai, sedangkan "is" digunakan untuk melihat identitas objek. Jadi, walaupun dua data memiliki nilai yang sama, belum tentu keduanya merupakan objek yang sama.

### 2.2.2 Program "assigment_2611533023.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
angka1_3023 = int(input("input angka-1_3023: "))
angka2_3023 = int(input("input angka-2_3023: "))

print("\nNilai awal angka1_3023 =", angka2_3023)
print("nilai angka2_3023 =", angka2_3023)

# Assignment biasa
hasil_3023 = angka1_3023
print("\nAssignment biasa (=)")
print("Hasil =", hasil_3023)

#Assignment penambahan
hasil_3023 = angka1_3023
hasil_3023+= angka2_3023
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_3023)

#Assignment pengurangan
hasil_3023 = angka1_3023
hasil_3023 -= angka2_3023
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil_3023)

#Assignment perkalian
hasil_3023 = angka1_3023
hasil_3023 *= angka2_3023
print("\nAssigment perkalian (*=)")
print("Hasil =", hasil_3023)

#assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_3023 != 0:
    hasil_3023 = angka1_3023
    hasil_3023 /= angka2_3023
    print("\nAssignment pembagian bulat (//=)")
    print("hasil_3023 =", hasil_3023)
    #Operator tambahan
    hasil_3023 = angka1_3023
    hasil_3023 //= angka2_3023
    print("\nAssignment pembagian bulat (//=)")
    print("hasil_3023 =", hasil_3023)
    hasil_3023 = angka1_3023
    hasil_3023 %= angka2_3023
    print("\nAssignment sisa bagi (%=)")
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka keduka tidak boleh bernilai 0.")

#Operator tambahan: assignment perpangkatan
hasil_3023 = angka1_3023
hasil_3023 **= angka2_3023
print("\nAssignment perpangkatan (**=)")
print("hasil_3023 =", hasil_3023)
```

**Penjelasan :**
1. Operator = digunakan untuk memasukkan atau menyalin nilai dari variabel angka1_3023 secara langsung ke dalam variabel hasil_3023 tanpa melakukan proses perhitungan.
2. +=: Menambahkan nilai hasil_3023 dengan angka2_3023, lalu langsung menyimpan hasil barunya ke dalam variabel hasil_3023 (singkatan dari hasil = hasil + angka2).

-=: Mengurangi nilai hasil_3023 dengan angka2_3023 dan langsung menyimpan hasil perhitungannya.

*=: Mengalikan nilai hasil_3023 dengan angka2_3023 dan langsung menyimpan hasilnya.
3. if angka2_3023 != 0: Memastikan angka kedua bukan nol untuk mengindari terjadinya error pembagian dengan nol (ZeroDivisionError).

/=: Membagi variabel hasil_3023 dengan angka2_3023 yang menghasilkan bilangan desimal (float), lalu menyimpan hasilnya.

//=: Membagi variabel hasil_3023 dengan angka2_3023 lalu membulatkan hasilnya ke bawah (floor division) dan menyimpan nilainya.

%=: Mencari sisa dari pembagian variabel hasil_3023 dengan angka2_3023 (modulo), lalu menyimpan sisa tersebut.
4. **=: Memangkatkan nilai hasil_3023 sebanyak nilai angka2_3023 (eksponen), lalu langsung menyimpan hasil akhirnya ke dalam variabel hasil_3023.




### 2.2.3 Program "bitwise_2611533023.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
print("\n===================================")
print("3. OPERATOR BITWISE")
print("===================================")

angka1_3023 = int(input("masukan angka bitwise-1:"))
angka2_3023 = int(input("masukan angka bitwise-2:"))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1_3023 =", angka1_3023, "| biner =", bin(angka1_3023))
print("angka2_3023 =", angka2_3023, "| biner =", bin(angka2_3023))

# Bitwise AND
hasil_3023 = angka1_3023 & angka2_3023
print("\nBitwise AND (&)")
print(angka1_3023, "&", angka2_3023, "=", hasil_3023)
print("biner hasil_3023 =", bin(hasil_3023))
print("biner hasil_3023 (8 bit) =", format(hasil_3023, "08b"))

#Bitwise OR
hasil_3023 = angka1_3023 | angka2_3023
print("\nBitwise OR (|)")
print(angka1_3023, "|", angka2_3023, "=", hasil_3023)
print("biner hasil =", bin(hasil_3023))
print("biner hasil (8 bit) =", format(hasil_3023, "08b"))

#Bitwise XOR
hasil_3023 = ~angka1_3023 ^ angka2_3023
print("\nBitwise XOR (^)")
print(angka1_3023, "^", angka2_3023, "=", hasil_3023)
print("Biner hasil =", bin(hasil_3023))
print("Biner hasil (8 bit) =", format(hasil_3023, "08b"))

#Bitwise NOT
hasil_3023 = ~angka1_3023
print("\nBitwise NOT (~)")
print("~", angka1_3023, "=", hasil_3023)
print("biner hasil =", bin (hasil_3023))
print("biner hasil (8 bit) =", format(hasil_3023, "08b"))

#bitwise gaser kiri
jumlah_geser_3023 = int(input("\nMasukan jumlah pergeseran bit: "))

hasil_3023 = angka1_3023 << jumlah_geser_3023
print("\nBitwise geser kiri (<<)")
print(angka1_3023, "<<", jumlah_geser_3023, "=", hasil_3023)
print("biner hasil =", bin(hasil_3023))
print("biner hasil (8 bit) =", format(hasil_3023, "08b"))

#Bitwise geser kanan
hasil_3023 = angka1_3023 >> jumlah_geser_3023
print("\nBitwise geser kanan (>>)")
print(angka1_3023, ">>", jumlah_geser_3023, "=", hasil_3023)
print("biner hasil =", bin(hasil_3023))
print("biner hasil (8 bit) =", format(hasil_3023, "08b"))
```

**Penjelasan :**
1. Perintah bin() digunakan untuk mengubah angka biasa yang dimasukkan pengguna menjadi deretan angka 0 dan 1, sehingga komputer bisa memproses operasi di level paling dasar.
2. Operator "DAN" (Bitwise AND - &)
Perintah format(..., "08b") dipakai agar komputer menampilkan hasilnya dalam wujud 8 digit angka biner rapi.
3. Operator "ATAU" (Bitwise OR - |)
 Hasilnya akan bernilai 1 jika ada minimal satu saja angka 1 di antara kedua angka yang dibandingkannya.
Hasilnya baru akan bernilai 0 jika kedua posisinya sama-sama bernilai 0.
4. Operator "BEDA" (Bitwise XOR - ^)
 Operator ^ bertugas memberikan nilai 1 hanya jika pasangan angka yang dibandingkan itu berbeda (yang satu 1, yang satu lagi 0). Kalau angkanya kembar (sama-sama 1 atau sama-sama 0), hasilnya malah 0.
Di dalam kode ini ada tanda ~ di depan angka pertama. Artinya, posisi angka 0 dan 1 pada angka pertama dibalik dulu sebelum disandingkan dengan angka kedua.
5. Operator "KEBALIKAN" (Bitwise NOT - ~)
Bertugas menukar nilai. Semua angka 1 diubah jadi 0, dan semua angka 0 diubah jadi 1.
Dalam hitungan angka biasa, efek dari pembalikan ini akan mengubah angka tersebut menjadi negatif, lalu nilainya ditambah satu (misalnya angka 5 diubah jadi -6).
6. Geser Kiri (<<) dan Geser Kanan (>>)
Program meminta pengguna menentukan berapa langkah pergeseran yang diinginkan.

Geser Kiri (<<): Dorong deretan angka 0 dan 1 ke arah kiri. Efeknya pada angka biasa sama seperti mengalikan angka tersebut dengan kelipatan dua.

Geser Kanan (>>): Dorong deretan angka 0 dan 1 ke arah kanan (angka paling ujung kanan bakal terbuang). Efeknya pada angka biasa sama seperti membagi angka tersebut dengan kelipatan dua.



### 2.2.4 Program "lainnya_2611533023.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
print("==================================")
print("1. OPERATOR KEANGGOTAAN")
print("==================================")

#input beberapa dsata yang dipisahkan dengan koma
input_data_3023 = input("masukan beberapa angka, pisahkan dengan koma: ")

# mengubah input menjadi list integer
data_3023 = [int(angka.strip()) for angka in input_data_3023.split(",")]

nilai_dicari_3023 = (input("masukan angka yang ingin dicari: "))

#Operator in
hasil_3023 = nilai_dicari_3023 in data_3023
print("\nOperator keanggotaan IN")
print(nilai_dicari_3023, "not in", data_3023, "=", hasil_3023)

print("\n===================================")
print("2. OPERATOR IDENTITAS")
print("=====================================")

# objek1 menggunakan list dari input pengguna
objek1_3023 = data_3023

# objek2 merujuk pada objek yang sama dengan objek1
objek2_3023 = objek1_3023

#objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_3023 = data_3023.copy()

print("objek1_3023 =", objek1_3023)
print("objek2_3023 =", objek2_3023)
print("objek3_3023 =", objek3_3023)

#Operator is
hasil_3023 = objek1_3023 is objek2_3023
print("\nOperator identitas IS")
print("objek2_3023 =", hasil_3023)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1_3023 is objek3_3023 =", objek1_3023 is objek3_3023)
print("objek1_3023 == objek3 =", objek1_3023 == objek3_3023)
```

**Penjelasan :**
1. Program meminta pengguna memasukkan beberapa angka sekaligus yang dipisahkan oleh tanda koma.
Teks masukan tersebut dipotong-potong berdasarkan koma (split(",")), dibersihkan dari spasi yang tidak perlu (strip()), lalu diubah menjadi angka bulat (int()) dan dikumpulkan ke dalam sebuah daftar angka (disebut list).

2. Kata kunci in bertugas mengecek apakah angka yang dicari ada di dalam daftar angka yang sudah dimasukkan sebelumnya.
Hasil pengecekan ini hanya ada dua kemungkinan: True (Benar, jika angkanya ketemu) atau False (Salah, jika angkanya tidak ada).

3. objek1_3023: Mengambil langsung daftar angka dari input awal.
objek2_3023 = objek1_3023: Tidak membuat daftar baru, melainkan hanya memberi nama panggil/label kedua untuk daftar angka yang sama. Bayangkan satu wadah fisik yang ditempeli dua label nama berbeda.
objek3_3023 = data_3023.copy(): Perintah .copy() membuat wadah baru yang benar-benar terpisah, meskipun isi di dalam wadahnya persis sama dengan daftar awal

4. Operator is bukan mengecek apakah isinya sama, melainkan mengecek apakah dua variabel tersebut menunjuk ke wadah/fisik memori yang sama persis.
Karena objek1 dan objek2 menunjuk wadah yang sama, hasilnya adalah True.

5. objek1_3023 is objek3_3023 (Hasil: False): Meskipun isi angkanya sama, komputer menganggap keduanya berbeda karena objek3 dibuat di wadah/memori baru yang terpisah.
objek1_3023 == objek3_3023 (Hasil: True): Tanda == hanya peduli pada isinya. Selama angka di dalamnya sama persis, hasilnya akan dianggap benar (True), tidak peduli wadahnya sama atau beda.



### 2.2.5 Program "logika_2611533023.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
a1_3023 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_3023 = input("input nilai boolean-2 (true/false): ").strip().lower()

print("\nA1 =", a1_3023)
print("A2 =", a2_3023)

# Konjungsi: bernilai true jika keduanya true
hasil_3023 = a1_3023 and a2_3023
print("\nKonjungsi (AND)")
print("A1_3023 and A2 =", hasil_3023)

#disjungsi: bernilai true jika salah satunya true
hasil_3023 =  a1_3023 or a2_3023
print("\nDisjungsi (or)")
print("A1_3023 or A2_3023 =", hasil_3023)

#negasi A1_3023: membalik nilai A1_3023
hasil_3023 = not a1_3023
print("\nNegasi A1_3023 (NOT)")
print("not A1_3023 =", hasil_3023)

#Negasi A2_3023: membalik nilai A2_3023
hasil_3023 = not a2_3023
print("\nNegasi A2_3023 (NOT)")
print("not A2 =", hasil_3023)

#XOR: bernilai true jika kedua nilai berbeda
hasil_3023 = a1_3023 != a2_3023
print("\nDisjungsi Eksklusif (XOR)")
print("A1_3023 XOR A2_3023", hasil_3023)
```

**Penjelasan :**
1. Menerima dan Memproses Input (Benar/Salah)
Program meminta pengguna mengetik tulisan "true" (benar) atau "false" (salah).

Perintah .strip().lower() dipakai untuk membersihkan spasi tak terpakai dan mengubah teks menjadi huruf kecil semua (jadi kalau diketik "True" atau "TRUE", komputer tetap paham).

Tanda == "true" bertugas mengubah teks masukan tersebut menjadi nilai logika asli Python: True jika pengguna mengetik "true", dan False jika mengetik selain itu.

2. Operator "DAN" (Konjungsi - and) 
Hasilnya baru dianggap True (benar) jika kedua nilai sama-sama bernilai True.

Jika salah satu saja bernilai False (atau dua-duanya False), maka hasil akhirnya langsung jadi False.

3. Operator "ATAU" (Disjungsi - or)
Hasilnya akan bernilai True (benar) jika ada minimal satu saja nilai yang bernilai True.

Hasilnya baru akan bernilai False jika kedua input sama-sama bernilai False.

4. Operator "KEBALIKAN" (Negasi - not)
Bertugas membalikkan keadaan.

Jika nilai awalnya True, setelah diberi not nilainya berubah jadi False. Sebaliknya, jika nilai awalnya False, hasilnya berubah jadi True.

5. Operator "BEDA PILIHAN" (XOR - !=)
Di dalam Python, logika XOR (Disjungsi Eksklusif) bisa dibuat dengan tanda "tidak sama dengan" (!=).

Hasilnya akan bernilai True hanya jika kedua nilai input berbeda (yang satu True dan yang satu lagi False). Kalau kedua nilainya kembar (sama-sama True atau sama-sama False), hasilnya adalah False.

2. Operator "Lebih Besar Dari" (>)
Memeriksa apakah angka pertama nilainya lebih besar secara mutlak dari angka kedua.

Bernilai True (benar) jika angka pertama memang lebih besar, dan False (salah) jika angka pertama sama atau lebih kecil dari angka kedua.

3. Operator "Lebih Kecil Dari" (<)
Memeriksa apakah angka pertama nilainya lebih kecil secara mutlak dari angka kedua.

Bernilai True jika angka pertama lebih kecil, dan False jika angka pertama lebih besar atau sama nilainya.

4. Operator "Lebih Besar atau Sama Dengan" (>=)
Memeriksa apakah angka pertama lebih besar ATAU setara dengan angka kedua.

Berbeda dengan > biasa, jika angka pertama dan angka kedua bernilai kembar (misal sama-sama 10), hasilnya tetap dianggap True.

5. Operator "Lebih Kecil atau Sama Dengan" (<=)
Memeriksa apakah angka pertama lebih kecil ATAU setara dengan angka kedua.

Hasilnya bernilai True selama angka pertama tidak melebihi angka kedua. Jika nilainya persis sama, hasilnya tetap True.

6. Operator "Sama Dengan" (==)
Memeriksa apakah kedua angka memiliki nilai yang persis sama.

Bernilai True jika nilainya kembar (misal 5 dan 5), dan bernilai False jika ada perbedaan angka sedikit pun. (Tanda == dipakai khusus untuk membandingkan, bukan untuk mengisi nilai).

7. Operator "Tidak Sama Dengan" (!=)
Kebalikan dari sama dengan. Operator ini memeriksa apakah kedua angka berbeda satu sama lain.

Bernilai True jika angkanya berbeda (misal 5 dan 7), dan bernilai False jika angkanya persis sama.

8. Perbandingan Berantai (Rentang Nilai)

Bernilai True jika angka tersebut bernilai di antara 1 sampai 99. Jika angkanya 0, 100, atau bernilai negatif, hasilnya jadi False.
