## 2.1 Dasar Teori
Praktikum ini membahas konsep dasar mengenai variabel, tipe data, dan konstanta dalam Python. Konsep tersebut merupakan bagian penting dalam pembuatan program karena digunakan untuk menyimpan, mengolah, dan memberikan nilai pada data. Adapun beberapa konsep yang dipelajari dalam praktikum ini adalah sebagai berikut. konsep-konsep yang digunakan waktu praktikum minggu kedua yaitu:

1. Variabel merupakan tempat untuk menyimpan data yang dapat digunakan dalam suatu program. Setiap variabel memiliki nama dan nilai yang diberikan kepadanya. Nilai tersebut dapat berupa angka, teks, maupun jenis data lainnya.
Dalam Python, penamaan variabel memiliki beberapa aturan. Nama variabel bersifat case-sensitive, sehingga penggunaan huruf besar dan huruf kecil dianggap berbeda. Contohnya, variabel_Ku dan variabel_ku merupakan dua variabel yang berbeda. Selain itu, nama variabel tidak boleh menggunakan kata kunci yang sudah digunakan dalam Python,seperti if, while, dan for.
2. Tipe data numerik digunakan untuk menyimpan data berupa angka. Tipe data ini dapat digunakan dalam berbagai perhitungan matematika, seperti penjumlahan, pengurangan, perkalian, dan pembagian. Contohnya adalah bilangan bulat dan bilangan desimal.
3. String merupakan tipe data yang digunakan untuk menyimpan data berupa teks atau kumpulan karakter. Data string ditulis menggunakan tanda petik. Dalam praktikum, tipe data string dapat digunakan untuk membuat biodata diri, seperti nama, tempat tinggal, jurusan, dan informasi lainnya.
4. Boolean merupakan tipe data yang hanya memiliki dua nilai, yaitu True dan False. Nilai tersebut dapat digunakan untuk menentukan suatu kondisi benar atau salah. Penulisan True dan False harus diawali dengan huruf kapital dan tidak menggunakan tanda petik.
Salah satu penerapannya adalah untuk menentukan hasil kelulusan atau status cumlaude. Program dapat memberikan nilai True apabila suatu kondisi terpenuhi dan False apabila kondisi tersebut tidak terpenuhi.
5. Input digunakan untuk menerima data yang diberikan oleh pengguna ketika program dijalankan. Data yang dimasukkan dapat digunakan untuk melakukan suatu proses atau perhitungan. Salah satu contohnya adalah program untuk menghitung luas persegi panjang, di mana pengguna memasukkan nilai panjang dan lebar, kemudian program menghitung luas berdasarkan data tersebut.
6. Konstanta merupakan nilai yang ditetapkan untuk digunakan dalam program dan pada dasarnya tidak dimaksudkan untuk diubah. Dalam Python, deklarasi konstanta dapat dilakukan dengan bantuan Final yang terdapat pada modul typing.
Untuk menggunakan Final, modul tersebut perlu diimpor terlebih dahulu menggunakan from dan import. Penggunaan konstanta dapat membantu memberikan penanda bahwa suatu nilai ditetapkan dan tidak seharusnya diubah selama program digunakan.

### 2.2.1 Program "Boolean_2611533023.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

"tugas2_2611533023.py"


```python
#nilai_3023
#Deklarasi variabel dengan tipe data boolean
is_lulus = True
is_cumlaude = True#nilai_3023
#Deklarasi variabel dengan tipe data boolean
is_lulus = True
is_cumlaude = True

#menggunakan Boolean
nilai = 85
batas_lulus = 75

#menentukan nilai Boolean dari kondisi
status_kelulusan = nilai >= batas_lulus  #hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai)
print("Apakah Lulus?:", status_kelulusan)
if is_lulus and is_cumlaude:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")

#menggunakan Boolean
nilai = 85
batas_lulus = 75

#menentukan nilai Boolean dari kondisi
status_kelulusan = nilai >= batas_lulus  #hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai)
print("Apakah Lulus?:", status_kelulusan)
if is_lulus and is_cumlaude:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

**Penjelasan :**
1. Inisialisasi Variabel Boolean
Pada baris awal, dilakukan deklarasi variabel is_lulus dan is_cumlaude. Kedua variabel ini secara eksplisit diberikan nilai literal True. Variabel ini nantinya berfungsi sebagai flag atau penanda kondisi spesifik mahasiswa dalam struktur kendali.
2. Operasi Perbandingan
Variabel nilai diisi dengan data integer 85, sedangkan variabel batas_lulus diisi dengan 75.
Dilakukan operasi perbandingan menggunakan operator >= (lebih besar dari atau sama dengan) untuk membandingkan kedua variabel tersebut.
3. Output Data
Blok ini berfungsi untuk menampilkan visualisasi data hasil pemrosesan ke console / terminal. Fungsi print() memanggil variabel nilai dan status_kelulusan yang bertipe data Boolean untuk diinformasikan kepada pengguna.
4. Struktur kendali logika
Baris ini menerapkan struktur percabangan if yang dikombinasikan dengan operator logika and.
Karakteristik dari operator and adalah hanya akan menghasilkan nilai True jika dan hanya jika kedua operand (variabel is_lulus dan is_cumlaude ) bernilai True.




### 2.2.2 Program "Input_2611533023.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
# panjang_3023
# Program ini menggunakan fungsi input() untuk meminta pengguna 
# memasukan nilai panjang dan lebar dari sebuah persegi panjang.
# Nilai yang dimasukan akan dikonversi menjadi tipe data integer
panjang_3023 = input('Masukan panjang persegi panjang:')
lebar_3023 = input('Masukan lebar persegi panjang:')
luas_3023 = int(panjang_3023) * int(lebar_3023)
print("luas =", luas_3023)
```

**Penjelasan :**
1. MemintaAngka Panjang & Lebar 
lewat perintah input(), komputer akan memunculkan teks dilayar: "masukan panjang persegi panjang" dan "masukan lebar persegi panjang"
2. Menghitung luas
Setelah  komputer tau angka pastinya, kompoter langsung mengalikan angka panjang dan angka lebar tersebut (int(panjang) * int(lebar)) sesuai rumus persegi panjang yaitu PxL
3. Menampilkan hasil
Terakhir yaitu, perintah print() bertugas untuk menampilkan hasilnya atau output




### 2.2.3 Program "konstanta_2611533023.py"


Jalankan sel kode di bawah ini agar Jupyter otomatis menyimpan kodenya beserta outputnya sebagai satu kesatuan (tidak perlu screenshot terpisah untuk hasil run).

```python
#jari_3023
from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_3023 = float(input('Masukan jari-jari lingkaran:'))
luas_3023 = PI * jari_3023 * jari_3023
print("luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3023, luas_3023))
```

**Penjelasan :**
1. Menyiapkan Rumus 
program mencatat rumus pertamanya yaaitu phi (3,14)
2. File Python
File Python merupakan tempat untuk menyimpan program yang telah dibuat. File Python memiliki ekstensi .py dan dapat dibuat serta diedit melalui VS Code.
3. Repository
Repository merupakan tempat penyimpanan suatu proyek yang dikelola menggunakan Git. Repository dapat berisi file program, dokumen, serta catatan perubahan yang berkaitan dengan proyek.
4. Commit
Commit merupakan proses untuk mencatat perubahan yang telah dilakukan pada suatu proyek. Setiap commit dapat diberikan keterangan agar perubahan yang dilakukan dapat diketahui dengan jelas.
5. Push
Push merupakan proses mengirimkan perubahan dari repository yang terdapat pada komputer ke repository yang terdapat di GitHub. Proses ini memungkinkan hasil pekerjaan yang telah dibuat dapat tersimpan pada GitHub.
6. Clone
Clone merupakan proses menyalin repository dari GitHub ke komputer. Dengan melakukan clone, pengguna dapat memperoleh salinan proyek yang tersimpan di GitHub untuk kemudian dikerjakan atau dikembangkan kembali.
